```python
## Happiness trend over Rank
df.replace([np.inf, -np.inf], np.nan, inplace=True)
import warnings
warnings.filterwarnings("ignore", message="use_inf_as_na option is deprecated")
df = pd.read_csv(r"C:\Users\GSDS\Desktop\DATASET\lifestyle.csv")
sns.lineplot(x='Rank', y='Happiness levels', data=df, marker='o')
plt.title("Happiness vs City Rank")
plt.xlabel("City Rank"); plt.ylabel("Happiness Score")
plt.gca().invert_xaxis(); plt.show()
```

```python
df.shape
```

```python
### Average Obesity by Happiness quartile
# Convert Obesity levels to numeric (removes %, commas, etc. if needed)
import warnings
warnings.simplefilter(action="ignore", category=FutureWarning)
grouped = df.groupby("Happiness quartile", observed=False)["Obesity levels"].mean().reset_index()

sns.barplot(x="Happiness quartile", y="Obesity levels", data=grouped, palette="Blues_d")
plt.title("Avg Obesity (%) per Happiness Quartile")
plt.ylabel("Obesity (%)")
plt.show()
```

```python
### Distribution of Sunshine hours
sns.histplot(df['Sunshine hours'], bins=15, kde=True, color='orange')
plt.title("Distribution of Annual Sunshine Hours")
plt.xlabel("Sunshine hours"); plt.show()
```

```python
sns.scatterplot(x='Pollution', y='Life expectancy', data=df, hue='City', s=100)
plt.title("Life Expectancy vs Pollution")
plt.xlabel("Pollution Index"); plt.ylabel("Life Expectancy (years)")
plt.legend(title="Continent"); plt.show()
```
